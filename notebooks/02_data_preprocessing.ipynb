```python
import pandas as pd
import numpy as np
import re
```

```python
# cargar el dataset
df = pd.read_csv('../data/raw/youtoxic_english_1000.csv')
df.head()
```

```python
# Comprobar duplicados
print("Filas duplicadas:")
print(f"Total de filas duplicadas: {df.duplicated().sum()}")
print(f"\nDuplicados en la columna 'Text': {df['Text'].duplicated().sum()}")

# Mostrar ejemplos de duplicados si existen
if df['Text'].duplicated().sum() > 0:
    print("\nEjemplos de textos duplicados:")
    duplicated_texts = df[df['Text'].duplicated(keep=False)].sort_values('Text')
    print(duplicated_texts[['Text', 'IsToxic']].head(10))
```

```python
# Eliminar columnas de ID (no aportan información al modelo)
print("Columnas antes de eliminar:", df.columns.tolist())
print(f"Shape antes: {df.shape}")
df_clean = df.drop(columns=['CommentId', 'VideoId'])

print(f"\nColumnas después de eliminar: {df_clean.columns.tolist()}")
print(f"Shape después: {df_clean.shape}")

df_clean.head()
```

```python
# Convertir la columna 'Text' a minúsculas
df_clean['Text'] = df_clean['Text'].str.lower()

print("\nEjemplo después de minúsculas:")

print(df_clean[['Text', 'IsToxic']].head(3))
```

```python
# Limpiar el texto: dejar solo letras y espacios, y normalizar espacios, tabs y saltos de línea a un solo espacio

def clean_text(text):
    """
    Reemplaza todo lo que no sea letra o espacio/tab/nueva línea por nada
    Reemplaza tabs y saltos de línea por espacio
    Reemplaza múltiples espacios por uno solo
    Elimina espacios al inicio y final
    """
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'[\t\n\r]+', ' ', text)
    text = re.sub(r' +', ' ', text)
    return text.strip()

df_clean['Text'] = df_clean['Text'].apply(clean_text)

print("Ejemplo de textos limpios y normalizados:")
print(df_clean[['Text', 'IsToxic']].head(3))
```

```python
# Guardar dataset preprocesado en data/processed/
df_clean.to_csv('../data/processed/youtoxic_clean.csv', index=False)

print(f"Dataset limpio guardado en: {'../data/processed/youtoxic_clean.csv'}")
print(f"Total de registros: {len(df_clean)}")
print(f"Columnas guardadas: {df_clean.columns.tolist()}")
```

## Conclusiones del preprocesamiento

- Se eliminaron columnas de ID irrelevantes para el modelado.
- El texto de los comentarios fue limpiado y normalizado: solo letras, minúsculas, sin signos ni espacios extra.
- No se eliminaron duplicados en la columna `Text`, ya que comentarios iguales pueden provenir de diferentes usuarios y cada fila representa una instancia válida.
- El dataset limpio se guardó en `data/processed/youtoxic_clean.csv` y está listo para el análisis y modelado.
