```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Configuración estética de las gráficas (Estilo profesional)
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# --- CONFIGURACIÓN DEL ARCHIVO ---
nombre_del_archivo = 'youtoxic_english_1000.csv'  

ruta_archivo = os.path.join('..', 'data', 'raw', nombre_del_archivo)

if os.path.exists(ruta_archivo):
    # 1. Cargar el dataset
    df = pd.read_csv(ruta_archivo)
    print(f"✅ ¡Éxito! Dataset cargado. Filas: {df.shape[0]}, Columnas: {df.shape[1]}")
    
    # 2. Vista previa rápida
    display(df.head())

    # 3. Limpieza preventiva (CRÍTICO)
    # Si hay filas vacías en el texto, darán error al contar palabras. Las eliminamos.
    nulos = df['text'].isnull().sum()
    if nulos > 0:
        print(f"⚠️ Se encontraron {nulos} filas vacías. Eliminándolas...")
        df = df.dropna(subset=['text'])
    else:
        print("✨ No se encontraron textos vacíos (nulos).")

else:
    print("\n" + "="*50)
    print("❌ ERROR CRÍTICO: NO SE ENCUENTRA EL ARCHIVO")
    print("="*50)
    print(f"El código buscó en: {os.path.abspath(ruta_archivo)}")
    print("SOLUCIÓN: Verifica que el nombre en 'nombre_del_archivo' sea idéntico al de tu carpeta data/raw/")
    # Definimos df como None para evitar errores posteriores, pero el código parará aquí lógicamente.
    df = None
```

Módulo 2: Análisis de Longitud de 
¿Por qué hacemos esto? Porque a menudo, los mensajes de odio tienen patrones de longitud diferentes (son más cortos y agresivos, o diatribas largas).

Paso 2.1: Crear métricas de longitud

Vamos a crear nuevas columnas numéricas. Esto es "Ingeniería de Características" básica.

```python
# Calcular número de caracteres
df['char_count'] = df['text'].astype(str).apply(len)

# Calcular número de palabras
df['word_count'] = df['text'].astype(str).apply(lambda x: len(x.split()))

# Ver los resultados
display(df[['text', 'char_count', 'word_count']].head())
```
