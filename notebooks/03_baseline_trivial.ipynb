# Baseline trivial: predicción de la clase mayoritaria

Este notebook implementa un modelo baseline muy simple que predice siempre la clase mayoritaria del dataset. Sirve como referencia mínima para comparar modelos más avanzados.

```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report
```

```python
# Cargar el dataset limpio
df = pd.read_csv('../data/processed/youtoxic_clean.csv')
df.head()
```

```python
# Ver la distribución de la variable objetivo
print(df['IsToxic'].value_counts())
df['IsToxic'].value_counts(normalize=True)
```

```python
# Obtener la clase mayoritaria
mayoritaria = df['IsToxic'].mode()[0]
print(f'Clase mayoritaria: {mayoritaria}')

# Predecir la clase mayoritaria para todas las filas
y_pred = np.full(shape=len(df), fill_value=mayoritaria)
y_true = df['IsToxic'].values
```

## 5. Evaluar el baseline con métricas básicas

```python
# Calcular métricas de evaluación
print('Accuracy:', accuracy_score(y_true, y_pred))
print('Precision:', precision_score(y_true, y_pred, zero_division=0))
print('Recall:', recall_score(y_true, y_pred, zero_division=0))
print('F1-score:', f1_score(y_true, y_pred, zero_division=0))

print('\nReporte de clasificación:')
print(classification_report(y_true, y_pred, zero_division=0))
```

## 6. Guardar las predicciones del baseline

## ¿Para qué sirve este baseline trivial?

Este baseline trivial predice siempre la clase mayoritaria, sin usar el texto. Sirve como referencia mínima: cualquier modelo real (que use el texto) debe superarlo en métricas como accuracy, precision, recall y F1-score. Si tu modelo no mejora este baseline, significa que no está aprendiendo nada útil y no merece la pena usarlo.

```python
# Guardar las predicciones en un CSV
df_baseline = df.copy()
df_baseline['baseline_pred'] = y_pred
df_baseline.to_csv('../data/processed/baseline_trivial_pred.csv', index=False)
print('Predicciones del baseline guardadas en ../data/processed/baseline_trivial_pred.csv')
```
